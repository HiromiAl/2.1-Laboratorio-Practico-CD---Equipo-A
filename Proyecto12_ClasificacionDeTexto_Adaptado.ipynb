# CASO ADAPTADO - CLASIFICACIÓN DE SPAM EN MENSAJES SMS

```python
import pandas as pd

# 1. Cargar el dataset de SMS directamente desde un repositorio público en GitHub
url = 'https://raw.githubusercontent.com/justmarkham/pycon-2016-tutorial/master/data/sms.tsv'
df_sms = pd.read_table(url, header=None, names=['label', 'message'])
```

```python
# Mostramos las primeras filas
print(df_sms.head())
print("\nConteo de clases:\n", df_sms['label'].value_counts())
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# 2. División de datos en Entrenamiento (67%) y Prueba (33%)
X_train, X_test, y_train, y_test = train_test_split(
    df_sms['message'], df_sms['label'], test_size=0.33, random_state=42
)
```

```python
# 3. Vectorización de Texto con TF-IDF
tfidf = TfidfVectorizer(stop_words='english')
X_train_vec = tfidf.fit_transform(X_train)
X_test_vec = tfidf.transform(X_test)
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# 4. Entrenamiento del modelo SVC
model_spam = SVC(kernel='linear')
model_spam.fit(X_train_vec, y_train)
```

```python
# 5. Predicción sobre el conjunto de prueba
y_pred_spam = model_spam.predict(X_test_vec)
```

```python
# 6. Evaluación Completa del Modelo
print("Accuracy (Precisión Global):", accuracy_score(y_test, y_pred_spam))
print("\n--- REPORTE DE CLASIFICACIÓN ---")
print(classification_report(y_test, y_pred_spam))
print("\n--- MATRIZ DE CONFUSIÓN ---")
print(confusion_matrix(y_test, y_pred_spam))
```

```python
# 7. Prueba
mensaje_nuevo = ["WINNER! You have won a $1,000 gift card. Call now to claim!"]
mensaje_vec = tfidf.transform(mensaje_nuevo)
print("Resultado para mensaje sospechoso:", model_spam.predict(mensaje_vec))
```
